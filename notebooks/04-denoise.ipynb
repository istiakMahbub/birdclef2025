# Denoising Exploration (Optional Step)

In this section, I experimented with **audio denoising** using the `noisereduce` library to clean the training clips before feature extraction. The motivation was to:

- Reduce background noise such as wind, rustling leaves.
- Improve model focus on core frequency components of species calls.

### Method:
- I applied noise reduction to each waveform using a simple noise profile estimated from the beginning portion of the audio.
- The denoised waveform was then passed through the same mel spectrogram and feature extraction pipeline.

### Why Denoising Was Excluded from Final Pipeline:

After testing, I decided **not to include denoising in the final training pipeline** due to the following observations:

1. **Real-world test data is noisy**: In practice, real bird call recordings often contain ambient sounds. Removing them during training may create a mismatch and reduce generalization.
2. **Potential loss of signal**: Aggressive denoising sometimes distorted faint but valid species signals.
3. **Lower performance**: Interestingly, training on *raw (un-denoised)* waveforms yielded **better validation accuracy** and **better test leaderboard performance**.

### Conclusion:

While denoising might help in some contexts, for **BirdCLEF 2025**, I chose to rely on the raw audio for better generalization and performance.


```python
import numpy as np
import IPython.display as ipd

# Load original waveform
npz_path = 'processed/raw_waveforms/CSA18786.npz'
data = np.load(npz_path)
y = data['waveform']
sr = int(data['sample_rate'])

# Play original audio
ipd.Audio(y, rate=sr)
```

```python
import noisereduce as nr
y_denoised = nr.reduce_noise(y=y, sr=sr)
```

```python
ipd.Audio(y_denoised, rate= sr)
```

```python
import os

def denoise_audio(signal, sr=32000, prop_decrease=1.0):
    """Denoise using noisereduce"""
    return nr.reduce_noise(y=signal, sr=sr, prop_decrease=prop_decrease)

# Directories
input_dir = 'processed/raw_waveforms'
output_dir = 'processed/denoised'
os.makedirs(output_dir, exist_ok=True)

# Loop through all .npz files
for fname in os.listdir(input_dir):
    if fname.endswith('.npz'):
        input_path = os.path.join(input_dir, fname)
        output_path = os.path.join(output_dir, fname)

        # Load waveform
        data = np.load(input_path)
        y = data['waveform']
        sr = data['sample_rate'].item() if isinstance(data['sample_rate'], np.ndarray) else data['sample_rate']

        # Denoise
        denoised_y = denoise_audio(y, sr)

        # Save
        np.savez_compressed(output_path, y=denoised_y, sr=sr)

        print(f"Denoised and saved: {fname}")
```
